### Start Training

Here we load the pre-trained `yolov8n.pt` and fine-tune it. 
The dataset YAML is located at `datasets/afternoon_v1/yolo_dataset/dataset.yaml`.

If you encounter path issues (because YOLO resolves relative paths from its current working directory), make sure you run this notebook from the main project root.

```python
# Load a pretrained YOLOv8 model
model = YOLO('yolov8n.pt')

# Define training arguments
dataset_yaml = r'datasets/afternoon_v1/yolo_dataset/dataset.yaml'

# Train the model
# We use imgsz=640 (standard) and epochs=50 to start. You can increase epochs to 100 later if needed.
results = model.train(
    data=dataset_yaml, 
    epochs=50, 
    imgsz=640, 
    batch=8,           # Lower batch size in case of limited local RAM/VRAM
    device='',         # Uses GPU if available, else CPU
    project='datasets/afternoon_v1/runs', 
    name='train_afternoon',
    exist_ok=True
)
```

### Validation and Results
After training, YOLO outputs the metrics to `datasets/afternoon_v1/runs/train_afternoon`. 
You can view the confusion matrix, F1-curve, and validation predictions inside that folder.

```python
# Validate the model
metrics = model.val()
print(f"mAP50-95: {metrics.box.map}")
print(f"mAP50: {metrics.box.map50}")
```
