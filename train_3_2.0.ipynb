# YOLOv8 Training: Crosswalk Pedestrian Detection (Merged V2)

**Dataset:** 1,111 labeled frames (Morning V1 + Afternoon V1 + Afternoon V2 combined)  
**Classes:** adult (0), child (1)  
**Architecture:** YOLOv8 Nano  
**Training:** 100 epochs, cosine LR, early stopping at patience=20

```python
# Cell 1: Setup
try:
    import ultralytics
except ImportError:
    !pip install ultralytics
    import ultralytics

from ultralytics import YOLO
ultralytics.checks()
```

## Start Training

All paths are absolute to avoid any working-directory confusion between Colab and local Jupyter.

```python
# Cell 2: Train
model = YOLO('yolov8n.pt')

DATASET_YAML = r'C:/Users/Acer/Documents/CAPSTONE/web_dashboard_analytics/datasets/merged_v2/data.yaml'
PROJECT_DIR  = r'C:/Users/Acer/Documents/CAPSTONE/web_dashboard_analytics/datasets/merged_v2/training_runs'

results = model.train(
    data=DATASET_YAML,
    epochs=100,
    imgsz=640,
    batch=8,
    workers=0,          # MUST be 0 on Windows
    device='',          # Auto-select GPU
    cache=True,         # Cache images in RAM for speed
    project=PROJECT_DIR,
    name='run_v2',
    exist_ok=True,
    patience=20,        # Early stop if no improvement for 20 epochs
    augment=True,
    cos_lr=True,        # Cosine LR schedule
    close_mosaic=20     # Disable mosaic in final 20 epochs
)
```

## Validate Best Model

After training completes, this cell loads the best checkpoint and runs validation.

```python
# Cell 3: Validate
best_pt = r'C:/Users/Acer/Documents/CAPSTONE/web_dashboard_analytics/datasets/merged_v2/training_runs/run_v2/weights/best.pt'
model = YOLO(best_pt)
metrics = model.val()
print(f'mAP50-95: {metrics.box.map}')
print(f'mAP50:    {metrics.box.map50}')
```
